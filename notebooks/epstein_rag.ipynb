# Setting Up and Checking what's in our storage

```python
import os
import sys
sys.path.append('../gemini-file-search')
import gemini_file_search as gfs
client = gfs.get_client()
```

```python
file_stores = gfs.list_file_search_stores(client)
```

```python
file_stores_dict = {}

for file_store in file_stores:
    file_stores_dict[file_store.model_dump()['display_name']] = file_store.model_dump()['name']

file_stores_dict
```

# Uploading data testing (will be removed)

```python
documents_list = []

for file_store_name in file_stores_dict.values():
    documents_list += gfs.list_documents_in_store(client, file_store_name)

uploaded_documents_names = [] # creating a void list

for i in range(len(documents_list)):
    uploaded_documents_names.append(documents_list[i].model_dump()['display_name']) #

print(uploaded_documents_names)

# Parser for files
file_counter = 0
pack_counter = 0
dataset_counter = 0
#added for tests

for dataset in os.listdir("../data/epstein_documents"):
    file_store_display_name = f"epstein_{dataset.lower()}"
    print("Working with",file_store_display_name)

    for pack in os.listdir(f"../data/epstein_documents/{dataset}"):
        print("Trying to upload pack",pack,"of it")
        for file_name in os.listdir(f"../data/epstein_documents/{dataset}/{pack}"):

            file_directory = f"../data/epstein_documents/{dataset}/{pack}/{file_name}"
            print('Trying to upload', file_directory)

            if file_name in uploaded_documents_names:
                print(file_name, "is already uploaded")
                continue
            else:
                print(file_name, "is not uploaded, uploading...")
                gfs.upload_file_to_store(client, file_directory, file_stores_dict[file_store_display_name])

            # file_counter += 1
            # if file_counter > 5: #Stopper
            #     break
        # pack_counter +=1
        # file_counter = 0
        # if pack_counter > 0: #Stopper
        #     break
    dataset_counter += 1
    pack_counter = 0
    if dataset_counter > 0: #Stopper
        break
```
