```python
#8. linguistic analysis
import re

def extract_words(name):
    if pd.isna(name):
        return []
    # Split by non-alphanumeric characters (keeps only words and numbers)
    # This splits on underscores, parentheses, spaces, etc.
    words = re.split(r'[^a-zA-Z0-9]+', str(name))
    # Filter out empty strings and convert to lowercase
    return [w.lower() for w in words if w]

# Create a temporary dataframe for processing
# We use .copy() to avoid SettingWithCopyWarning
temp_df = df[['name', 'count','type']].copy()
temp_df['word'] = temp_df['name'].apply(extract_words)

# Explode the list of words so each word gets its own row
exploded = temp_df.explode('word')

# Group by word and calculate statistics
word_stats = exploded.groupby(['word','type']).agg(
    tag_count=('name', 'nunique'),      # Count how many tags contain this word
    posts_count=('count', 'sum')      # Sum of posts for tags containing this word
).reset_index()

# Sort by tag_count to find the most prevalent words in names
word_stats = word_stats.sort_values(by='tag_count', ascending=False)

# Display the top 20 words
word_stats.head(20)
```
